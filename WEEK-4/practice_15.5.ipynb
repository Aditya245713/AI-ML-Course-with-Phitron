### 1. Add New Column and Save File
### You have a DataFrame with column Price.
### Create a new column Discounted_Price with (10% discounted price).
### Save the modified DataFrame as discounted_products.csv.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO

data = """
EmployeeID,EmployeeName,Age,Salary,Price,Marks1,Marks2,Marks3
E101,Alice,25,50000,1200,85,90,88
E102,Bob,30,60000,1500,78,82,80
E103,Charlie,22,52000,1100,92,95,90
E101,Alice,25,50000,1200,85,90,88
E104,David,45,80000,2500,88,85,89
E105,Eve,29,58000,1300,76,79,81
E106,Frank,,75000,2200,90,91,92
E107,Grace,34,,1900,81,83,80
E108,Heidi,38,,1800,84,86,88
E109,Ivan,26,53000,1400,95,92,94
E102,Bob,30,60000,1500,78,82,80
"""

df = pd.read_csv(StringIO(data))
df['Discounted_Price'] = df['Price']*0.9
df.to_csv('discounted_products.csv', index = False)
df_2 = pd.read_csv('discounted_products.csv')
df_2
```

### 3. Handle Missing Values
### Columns: Age, Salary
### Tasks:
### a) Count missing values
### b) Fill missing Age with mean age
### c) Drop rows where Salary is missing

```python
df
```

```python
print(df['Age'].isnull().sum())
print(df['Salary'].isnull().sum())
```

```python
df.fillna({'Age': df['Age'].mean()}, inplace=True)
df
```

```python
df.dropna(subset=['Salary'], inplace=True)
df
```

### 4. Apply Statistical Functions
### Columns: Marks1, Marks2, Marks3
### Tasks:
### a) Calculate mean, max for each subject
### b) Compute total marks per student
### c) Find standard deviation of total marks

```python
print(round(df['Marks1'].mean(), 2))
print(df['Marks1'].max())
print(round(df['Marks2'].mean(), 2))
print(df['Marks2'].max())
print(round(df['Marks3'].mean(), 2))
print(df['Marks3'].max())
```

```python
df['Total Marks'] = df.iloc[::, 5:8].sum(axis = 1)
df
```

```python
print(df['Total Marks'].std())
```

### 5. Histogram, Bar, Scatter, Pie
### Create:
### a) Histogram of Age
### b) Scatter plot (Age vs Salary)
### c) Scatter plot ( Price vs Discounted Price) 

```python
plt.hist(df['Age'], color='orange', edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Count')
plt.title('Histogram of Ages')
plt.show()
```

```python
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Scatter Plot of Age Vs Salary')
plt.scatter(df['Age'], df['Salary'], c='red')
plt.show()
```

```python
plt.xlabel('Price')
plt.ylabel('Discounted_Price')
plt.title('Scatter Plot of Price Vs Discounted_Price')
plt.scatter(df['Price'], df['Discounted_Price'], c='purple')
plt.show()
```
