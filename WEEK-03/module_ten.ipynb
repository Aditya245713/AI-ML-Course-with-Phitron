### NDARRAY

```python
import numpy as np 

# 1d array 
arr1 = np.array([1,2,3,4,5])

# 2d array 
arr2 = np.array([ [1,2,3], # row1
                  [4,5,6] #row2
                ])

#3d array 
arr3 = np.array([
    #1st floor
    [ [1,2,3], # row1
      [4,5,6]  ], #row2 

    #2nd floor
     [[1,2,3], # row1
      [4,5,6]  ], #row2  

    #3rd floor
      [[1,2,3], # row1
      [4,5,6]  ] #row2  
])

print(arr1)
print()
print(arr2)
print()
print(arr3)
```

### ATTRIBUTES

```python
# dimension 
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

print()

#shape 
print(arr1.shape)
print(arr2.shape)
print(arr3.shape) 

print()

#data type 
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

print()

#size 
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

### DATA TYPE

```python
arr = np.array([1, 2, 3]) 
print(arr.dtype)

arr = np.array([1.5, 2.4, 3.2]) 
print(arr.dtype)

#upcasting
arr = np.array([1, 2, 3.2]) 
# upcasted to float 
print(arr.dtype) 

#upcasted to string
arr = np.array([1, 2, 3.2, 'hello']) 
print(arr.dtype) 

arr = np.array([1, True, 3.25, 'hello']) # can't keep a list in it
print(arr.dtype)
```

### NDARRAY FROM EXISTING DATA

```python
# LIST
lst = [10,20,30,40 , 40.5] 
arr = np.array(lst) 
print(arr) 
print(type(arr)) 
print(arr.dtype) 

print()

mixed_lst = [10, True, 'Hello']
arr = np.array(mixed_lst, dtype = 'U32') 
print(arr) 
print(type(arr)) 
print(arr.dtype) 

print()

matrix = [ [1,2,3], #row1
           [4,5,6] #row2
        ]
arr = np.array(matrix) 
print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim) 
```

```python
#TUPLE
tpl = (10,20,30) 
arr = np.array(tpl,dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype) 
arr = arr.astype(np.int16) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
```

```python
#SET
st = { 1, 2, 3 }
arr = np.array(st) #takes whole set as object
# print(arr.dtype) 
# arr = arr.astype(np.int16)
print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)

print()

arr = np.array(list(st)) 
arr = arr.astype(np.int16)
print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim) 
```

```python
#DICTIONARY 
dc = { 'a' : 10 , 'b' : 20, 'c' : 30 }
keys = dc.keys()
values = dc.values()
items = dc.items()

#keys to ndarray
arr = np.array(list(keys))
print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
print(arr.size)

print()

#values to ndarray
arr = np.array(list(values))
print(arr) 
arr = arr.astype(np.int16)
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
print(arr.size)

print()

#items to ndarray
arr = np.array(list(items))
print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
print(arr.size)
```

### NDARRAY FROM SCRATCH

```python
# np.zeros(dimension, datatype)
arr = np.zeros((2,3),dtype=np.int8)

print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
print(arr.size)
print()

arr = np.zeros_like(arr3) ## arr3 shape same 
print(arr)
print(arr.shape)
```

```python
#np.ones(dimensions, datatype)
arr = np.ones((3,4,3),dtype=np.int8)

print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
print(arr.size)
print()

arr = np.ones_like(arr3) ## arr3 shape same 
print(arr)
print(arr.shape)
```

```python
#np.empty
arr = np.empty((4,3),dtype=np.int8)

print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
print(arr.size)
print()

arr = np.empty_like(arr3) ## arr3 shape same 
print(arr)
print(arr.shape)
```

```python
#np.full 
arr = np.full((4,3),np.inf)

print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
print(arr.size)
print()

arr = np.full_like(arr3,np.inf,dtype=np.float64) 
print(arr)
print(arr.shape)
print(arr.dtype)
```

### NDARRAY FROM RANGE

```python
# np.arange(first , end, step) -> similar to range function 
arr = np.arange(1 , 10, 1)
mat = arr.reshape(3,3) 
print(arr) 
print(mat) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
print(arr.size)
```

```python
# np.linspace(start , end(included), koita) 
arr = np.linspace(0, 4, 6)
print(arr) 
print(type(arr)) 
print(arr.dtype) 
print(arr.shape) 
print(arr.ndim)
print(arr.size)
```

```python
# np.logspace(start , end(included) , koita , base) 
arr = np.logspace(0, 4, 6, base = 2) #[2^0, 2^0.8, 2^1.6, 2^2.4, 2^3.2, 2^4]
print(arr)
```

### INDEXING AND SLICING

```python
#indexing 
#print(arr1)
#print(arr2)

# 1d array 
arr1 = np.array([1,2,3,4,5])

# 2d array 
arr2 = np.array([ [1,2,3], # row1
                  [4,5,6] #row2
                ])

#arr1[2] = 100 
#print(arr1)
arr2[0][2]= 100
print(arr2)
```

```python
#slicing 
#1d = arr[start:end:step] 

arr1_mod = arr1[1:4].copy() #eta korle arr1 er ekta independent copy create hoy
arr1[1] = 2                 #arr1_mod1 change korle arr1 change hobe na

arr1_mod[2] = 200
print(arr1_mod)
print(arr1)
```

```python
#2d slicing 
# arr[ row_start : row_end : step , column_start : column_end : step ]
# print(arr2) 

#getting a row
row_0 = arr2[0:1 ,]
row_1= arr2[1: , ]
print(row_0)
print(row_1)
print()

#getting a column 
col_0 = arr2[:: ,0:1]
col_1 = arr2[::,1:2]
print(col_0)
print(col_1)
print()

#getting a portion 
portion = arr2[:: , 0:2] 
print(portion)
```
