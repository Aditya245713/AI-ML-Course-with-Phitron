## 1. Students Marks Analyzer

```python
#You are helping a teacher check exam results.
# Create a NumPy array with 10 random integers between 0 and 100 (student scores).

# Find:
# - the highest score
# - the lowest score
# - the average score

# Hint: use np.random.randint(), arr.max(), arr.min(), arr.mean()

import numpy as np

stud_scores = np.random.randint(0, 100, 10)

print(stud_scores)
print()

highest_score = np.max(stud_scores)
lowest_score = np.min(stud_scores)
average_score = np.mean(stud_scores)

print(f"Highest score: {highest_score}, Lowest score: {lowest_score}, Average score: {average_score}")
```

## 2. Warehouse Weights

```python
#A warehouse stores item weights (in kg):
# arr = np.array([45, 60, 55, 30, 90, 75, 40, 50])

# Find and print:
# - all items heavier than 50 kg
# - the sorted weights
# - the index of the heaviest item

# Hint: use arr[arr > 50], np.sort(arr), arr.argmax()

arr = np.array([45, 60, 55, 30, 90, 75, 40, 50])

heavy_50 = arr[arr > 50]
sorted_arr = np.sort(arr)
heaviest = np.argmax(arr)

print(f"Heavier than 50 kg: {heavy_50}, Sorted array: {sorted_arr}, Heaviest weight index: {heaviest}")
```

## 3. Temperature Report

```python
#A weather station collects temperatures (°C) from 4 zones.
# Generate random integers between 20 and 40 for a 4×4 grid.

# Then find:
# - the average temperature of each zone (row)
# - the highest temperature in the grid
# - replace all temperatures above 35 with 35

temp = np.random.randint(20, 40, (4, 4))

print(temp)
print()

row_1 = temp[0:1,]
row_2 = temp[1:2,]
row_3 = temp[2:3,]
row_4 = temp[3:,]
print(f"Highest of 1st zone: {np.max(row_1)}, Highest of 2nd zone: {np.max(row_2)}, Highest of 3rd zone: {np.max(row_3)}, Highest of 4th zone: {np.max(row_4)}")
print()

highest_temp = np.max(temp)
print(f"Highest temperature in the grid: {highest_temp}")
print()

temp[temp > 35] = 35
print(temp)
```

## 4. Traffic Data

```python
#A city records the number of cars passing 5 intersections in 3 time slots (morning, afternoon, night).
# Create a 3×5 array of random integers between 0 and 50.

# Then find:
# - total cars per intersection (sum by column)
# - total cars per time slot (sum by row)

cars = np.random.randint(0, 50, (3, 5))
print(cars)
print()

inter_1 = cars[::, 0:1]
inter_2 = cars[::, 1:2]
inter_3 = cars[::, 2:3]
inter_4 = cars[::, 3:4]
inter_5 = cars[::, 4:]

print(f"Highest in intersection 1: {np.max(inter_1)}, intersection 2: {np.max(inter_2)}, intersection 3: {np.max(inter_3)}, intersection 4: {np.max(inter_4)}, intersection 5: {np.max(inter_5)}")
print()

time_1 = cars[0:1,]
time_2 = cars[1:2,]
time_3 = cars[2:,]

print(f"Highest in time slot 1: {np.max(time_1)}, time slot 2: {np.max(time_2)}, time slot 3: {np.max(time_3)}")
```

## 5. Fruit Store Stock

```python
#You sell fruits in your store. Stock levels (kg) for 5 days are:
# stock = np.array([12, 15, 9, 20, 18])

# A new delivery adds 5 kg to each stock.
# Then 10% of all fruits get spoiled.

# Find the final stock levels.

# Hint: use broadcasting and simple math.

stock = np.array([12, 15, 9, 20, 18])
new_del = stock + 5
spoilt = new_del * (1 - 0.1)

print(spoilt)
```

## 4. Filtering Employee Data

## 5. City Weather Dashboard

```python
data = {
 	'City': ['Dhaka', 'Sylhet', 'Chittagong', 'Rajshahi', 'Khulna'],
 	'Temp': [32, 28, 30, 35, 33],
 	'Humidity': [60, 75, 70, 55, 65]
}
df = pd.DataFrame(data)

# Tasks:
# 1. Show only the cities where FeelsLike > 33.
# 2. Sort the DataFrame by FeelsLike (descending).
# 3. Find which city has the highest humidity.

#FeelsLike = 𝑇 + (𝐻 - 50) / 5 (From gpt)

df['FeelsLike'] = df['Temp'] + (df['Humidity'] - 50) / 5
df.loc[df['FeelsLike'] > 33].drop('FeelsLike', axis = 1)
```

```python
df.sort_values(by = 'FeelsLike', ascending = 0).drop('FeelsLike', axis = 1)
```

```python
hum = df.sort_values(by = 'Humidity')
hum.tail(1).drop('FeelsLike', axis = 1)
```
