```python
#1. Create a Pandas Series from the list [10, 20, 30, 40, 50]. 
#Print the Series, its index, and data type. 
#Then create a DataFrame with columns "Name", "Age", and "City" containing at least 4 rows of data.

import pandas as pd

list = [10, 20, 30, 40, 50]
series = pd.Series(list, index = ['i', 'ii', 'iii', 'iv', 'v'])

print(series)
print(series.index)
print(type(series))
```

```python
#2. Using any DataFrame, 
#write code to print the number of rows and columns, column data types (.dtypes), 
#and summary statistics (.describe()).

new_dict = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Denver', 'Emily', 'Fiona', 'Gisele', 'Howard', 'Ilsa', 'Julia'],
    'Age': [25, 29, 28, 25, 23, 28, 29, 31, 25, 24],
    'City': ['Boston', 'Miami', 'Iowa', 'Florida', 'Pittsburgh', 'Miami', 'Washington', 'Florida', 'Vegas', 'Miami'],
    'CGPA': [3.94, 3.83, 3.77, 3.92, 3.75, 3.86, 3.94, 3.78, 3.84, 3.96]
}

new_df = pd.DataFrame(new_dict)

new_df.info()
print()
row = len(new_df.index)
col = len(new_df.columns)
print(f"ROWS: {row}, COLUMNS:{col}")
print()
print(new_df.dtypes)
print()
new_df.describe()
```

```python
#3. Create a DataFrame using the dictionary below:¶
#data = {
#'Product': ['Pen', 'Book', 'Eraser', 'Pencil'],
#'Price': [10, 50, 5, 7],
#'Quantity': [100, 60, 200, 150]
#}
#Print .columns, .index. Access a single column and one full row.

data = {
    'Product': ['Pen', 'Book', 'Eraser', 'Pencil'],
    'Price': [10, 50, 5, 7],
    'Quantity': [100, 60, 200, 150]
}
data_df = pd.DataFrame(data, index=[1, 2, 3, 4])

print(data_df.index)
print()
print(data_df.columns)
print()
print(data_df['Product'])
print()
print(data_df.loc[0:1])
```

```python
#4. Using .loc[] and .iloc[], show the first row, the second column, 
#and a 2×2 block from the middle of your DataFrame.

data = {
    'Code': ['P-001', 'P-002', 'P-003', 'P-004'],
    'Product': ['Pen', 'Book', 'Eraser', 'Pencil'],
    'Price': [10, 50, 5, 7],
    'Quantity': [100, 60, 200, 150]
}
data_df = pd.DataFrame(data)
data_df.set_index('Code', inplace=True)

print(data_df)
print()
print(data_df.iloc[0:1])
print()
print(data_df.iloc[::,1:2])
print()
print(data_df.iloc[1:3,0:2])
```

```python
#5. Create a 4×3 DataFrame of student names and marks. 
#Drop one column using .drop(), drop one row by index

my_tuple = (('Alice', 24), ('Bob', 30), ('Charlie', 28), ('Denver', 26))
tuple_df = pd.DataFrame(my_tuple,columns=['Name', 'Marks'], index = [1, 2, 3, 4])
new_tup_df = tuple_df.drop(3).drop('Name', axis = 1)
new_tup_df
```

```python
#6. Create a DataFrame of employees with columns "Name", "Department", and "Salary". 
#Sort it by "Salary" in ascending order and by "Name" in descending order.

emp = {
    'Name': ['Rahim', 'Karim', 'Jodu', 'Modu', 'Kodu'],
    'Department': ['Comp. Sci.', 'Physics', 'Biology', 'Music', 'Math'],
    'Salary': [60000, 40000, 50000, 40000, 50000]
}

emp_df = pd.DataFrame(emp, index=[1, 2, 3, 4, 5])
emp_df.sort_values(['Salary', 'Name'], ascending=[1, 0], inplace=True)
emp_df
```

```python
#7. Create a DataFrame of 8 students with columns "Name" and "Score". 
#Filter and print students with Score > 60 and those with Score < 40.

stud = {
    'Name': ['Rahim', 'Karim', 'Jodu', 'Modu', 'Kodu', 'Goru', 'Sagol', 'Gadha', 'Hash', 'Murgi'],
    'Score': [24, 36, 76, 98, 26, 43, 87, 91, 55, 67]
}

stud_df = pd.DataFrame(stud)
print(stud_df.loc[stud_df['Score'] > 60])

print()
print(stud_df.loc[stud_df['Score'] < 40])
```
