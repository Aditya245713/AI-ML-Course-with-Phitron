```python
# You must run this code and use that data set for all the questions
import numpy as np
import pandas as pd

np.random.seed(42)

ids = np.arange(1, 11)
ages = np.random.randint(18, 60, 10)
salaries = np.random.randint(30000, 90000, 10)
departments = np.array(["HR", "IT", "Finance", "IT", "HR", "Sales", "Finance", "IT", "Sales", "HR"])

DF = pd.DataFrame({
"id": ids,
"age": ages,
"salary": salaries,
"dept": departments
})

DF.to_csv("employees.csv", index=False)
print("Sample Data Created and Saved as employees.csv")
```

**Answer to Question: 1.**

```python
#Q1. Create NumPy arrays from the ages and salaries data generated above.
#Print both arrays and display their dtype, ndim, shape, and size.

age_arr = np.array(ages)
sal_arr = np.array(salaries)

print(age_arr)
print(f"Age dtype: {age_arr.dtype}")
print(f"Age ndim: {age_arr.ndim}")
print(f"Age shape: {age_arr.shape}")
print(f"Age size: {age_arr.size}")

print()

print(sal_arr)
print(f"Salary dtype: {sal_arr.dtype}")
print(f"Salary ndim: {sal_arr.ndim}")
print(f"Salary shape: {sal_arr.shape}")
print(f"Salary size: {sal_arr.size}")
```

**Answer to Question: 2.**

```python
#Q2. Using the salaries array, find and print the highest and lowest salary values.
#Also, calculate and print the average salary and age using NumPy functions.

max = np.max(sal_arr)
min = np.min(sal_arr)
avg_sal = np.mean(sal_arr)
avg_age = np.mean(age_arr)

print(f"Highest salary: {max}, Lowest salary: {min}, Average salary: {avg_sal}, Average age: {avg_age}]")
```

**Answer to Question: 3.**

```python
#Q3. From the ages array, filter and print all ages greater than 30 using a boolean condition.
#Then count and print how many employees are older than 30.

greater_30 = age_arr[age_arr > 30]
print(greater_30)
print()

print(f"No. of employees older than 30: {len(greater_30)}")
```

**Answer to Question: 4.**

```python
#Q4. Create a new NumPy array that increases every employee’s age by 5 years (without modifying the original array).
#Print the new updated ages array.

incage_5 = age_arr + 5
print(f"New Ages Array: {incage_5}")
print()

print(f"Original Array: {age_arr}")
```

**Answer to Question: 5.**

```python
#Q5. Using NumPy, calculate the total salary expense (sum of all salaries)
#and the difference between the maximum and minimum salary. Print both results.

print(f"Total salary expense: {np.sum(sal_arr)}")
print()

print(f"Difference between max and min salary = {max - min}")
```

**Answer to Question: 6.**

```python
#Q6. Load the employees.csv file into a Pandas DataFrame.
#Display the first 5 rows, and check basic info using info() and summary statistics using describe().

empdf = pd.read_csv('employees.csv')

print(empdf.head(5))
print()

print(empdf.info())
print()

print(empdf.describe())
```

**Answer to Question: 7.**

```python
#Q7. Display only the id, age, and salary columns.
#Then show the last 3 rows using tail().

print(empdf.loc[:,['id', 'age', 'salary']])
print()

print(empdf.tail(3))
```

**Answer to Question: 8.**

```python
#Q8. Filter the DataFrame to show only employees who work in the IT department.
#Print the result and show the total number of IT employees.

short_empdf = empdf.loc[empdf['dept'] == 'IT']
print(short_empdf)

emp_no = len(short_empdf.index)
print()
print(f"Total number of IT employees: {emp_no}")
```

**Answer to Question: 9.**

```python
#Q9. Sort the DataFrame by salary in descending order
#and display the top 3 highest-paid employees along with their department and age.

sal_empdf = empdf.sort_values('salary', ascending = 0)
print(sal_empdf.loc[:, ['id', 'dept', 'age']].head(3))
```

**Answer to Question: 10.**

```python
#Q10. Replace all salary values greater than 80000 with 80000 using loc.
#Then calculate and print the new average salary of all employees after replacement.

empdf.loc[empdf['salary'] > 80000, 'salary'] = 80000

print(f"New average salary of employees: {empdf['salary'].mean()}")
```
